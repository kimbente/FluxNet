```python
import torch
from models import FluxNet, DualBranchFluxNet

def print_param_count(model, model_name):
    # Sum up the number of elements (numel) for every trainable parameter
    total_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"{model_name} Total Trainable Parameters: {total_params:,}")

# Instantiate both models with default settings
flux_net = FluxNet()
dual_branch_flux_net = DualBranchFluxNet()

# Print and compare
print_param_count(flux_net, "FluxNet (Original)")
print_param_count(dual_branch_flux_net, "DualBranchFluxNet (Split)")
```
